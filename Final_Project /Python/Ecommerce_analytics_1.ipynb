### Seller city cleaning

```python
Orders_Sellers['seller_zip_code_prefix'] = Orders_Sellers['seller_zip_code_prefix'].astype(str).str.zfill(5)
Orders_Sellers['seller_zip_code_prefix'].str.len().value_counts()
```

```python
Orders_Sellers[Orders_Sellers['seller_city'].str.contains(r"[^a-z ]|  ")][['seller_city', 'seller_zip_code_prefix', 'seller_state']]
```

```python
import re
import unicodedata

def fold(text):
    text = str(text).replace("´", "'").replace("`", "'").replace("’", "'")
    text = unicodedata.normalize("NFKD", text).encode("ascii", "ignore").decode()
    return re.sub(r"\s+", " ", text).strip().lower()

# most common customer city for each zip code
Zip_To_City = (Customers.groupby(['customer_zip_code_prefix', 'customer_city']).size()
               .reset_index(name='n').sort_values('n', ascending=False)
               .drop_duplicates('customer_zip_code_prefix')
               .set_index('customer_zip_code_prefix')['customer_city'])

Seller_Typos = {
    "sao paulop": "sao paulo", "sao paluo": "sao paulo", "sao pauo": "sao paulo",
    "ao bernardo do campo": "sao bernardo do campo", "sao bernardo do capo": "sao bernardo do campo",
    "balenario camboriu": "balneario camboriu", "belo horizont": "belo horizonte",
    "cascavael": "cascavel", "floranopolis": "florianopolis", "garulhos": "guarulhos",
    "juzeiro do norte": "juazeiro do norte", "mogi das cruses": "mogi das cruzes",
    "paincandu": "paicandu", "portoferreira": "porto ferreira",
    "ribeirao pretp": "ribeirao preto", "riberao preto": "ribeirao preto", "robeirao preto": "ribeirao preto",
    "s jose do rio preto": "sao jose do rio preto", "sao jose do rio pret": "sao jose do rio preto",
    "sando andre": "santo andre", "sao jose dos pinhas": "sao jose dos pinhais",
    "scao jose do rio pardo": "sao jose do rio pardo", "tabao da serra": "taboao da serra",
    "aguas claras df": "aguas claras",
    "embu guacu": "embu-guacu", "ji parana": "ji-parana", "mogi guacu": "mogi-guacu",
}

# values that are not a city name: take the city from the zip code
Not_A_City = {"sp", "sbc", "bahia", "minas gerais", "santa catarina", "centro", "pirituba", "vicente de carvalho"}

def clean_seller_city(city, state, zip_code):
    c = fold(city)
    c = re.sub(r"\(.*?\)", "", c)                 # remove text in brackets
    c = re.split(r"[/\\,]", c)[0].strip()         # keep the part before / \ or ,
    st = str(state).lower()
    c = re.sub(r"\s*-\s*" + st + r"$", "", c)     # trailing '- sp'
    c = re.sub(r"\s+" + st + r"$", "", c)         # trailing ' sp'
    c = c.replace(" d oeste", " d'oeste")
    c = Seller_Typos.get(c, c)
    if c in Not_A_City or c.isdigit() or "@" in c or c == "":
        c = Zip_To_City.get(zip_code, c)
    return c
```

```python
Sellers_Before = Orders_Sellers.copy()

Orders_Sellers['seller_city'] = [clean_seller_city(c, s, z) for c, s, z in
                                 zip(Orders_Sellers['seller_city'], Orders_Sellers['seller_state'], Orders_Sellers['seller_zip_code_prefix'])]

Changed = Orders_Sellers['seller_city'] != Sellers_Before['seller_city']
Changed.sum()
```

```python
Sellers_Before.loc[Changed, ['seller_city', 'seller_zip_code_prefix', 'seller_state']].assign(new_city=Orders_Sellers.loc[Changed, 'seller_city'])
```

```python
# no rows were deleted (the translation table gets the 2 missing categories)
for name in Rows_After:
    expected = Rows_Before[name] + (2 if name == 'Orders_Products_Category' else 0)
    assert Rows_After[name] == expected, f"{name}: {Rows_After[name]} rows, expected {expected}"

# keys
assert Customers['customer_id'].is_unique
assert Orders['order_id'].is_unique
assert Orders_Products['product_id'].is_unique
assert Orders_Sellers['seller_id'].is_unique
assert Orders_Products_Category['product_category_name'].is_unique
assert not Orders_Items.duplicated(['order_id', 'order_item_id']).any()
assert not Orders_Payments.duplicated(['order_id', 'payment_sequential']).any()
assert not Orders_Reviews.duplicated(['review_id', 'order_id']).any()

# foreign keys (no orphans)
assert Orders['customer_id'].isin(Customers['customer_id']).all()
assert Orders_Items['order_id'].isin(Orders['order_id']).all()
assert Orders_Items['product_id'].isin(Orders_Products['product_id']).all()
assert Orders_Items['seller_id'].isin(Orders_Sellers['seller_id']).all()
assert Orders_Payments['order_id'].isin(Orders['order_id']).all()
assert Orders_Reviews['order_id'].isin(Orders['order_id']).all()

# cleaning was applied
assert Customers['customer_zip_code_prefix'].str.len().eq(5).all()
assert Orders_Sellers['seller_zip_code_prefix'].str.len().eq(5).all()
assert not Orders_Sellers['seller_city'].str.contains(r"[/\\,@]").any()
assert (Orders_Payments['payment_installments'] == 0).sum() == 0
assert (Orders_Products['product_weight_g'] == 0).sum() == 0
assert Orders_Products['product_weight_g'].isna().sum() == 6
assert Orders_Products['product_category_name'].isna().sum() == 610
assert Orders_Products['product_category_name_english'].notna().all()
assert not (Orders_Reviews['review_comment_title'].notna() & (Orders_Reviews['review_comment_title'].astype(str).str.strip() == '')).any()
assert not (Orders_Reviews['review_comment_message'].notna() & (Orders_Reviews['review_comment_message'].astype(str).str.strip() == '')).any()
for df, col in [(Orders, 'order_purchase_timestamp'), (Orders_Items, 'shipping_limit_date'), (Orders_Reviews, 'review_creation_date')]:
    assert str(df[col].dtype).startswith('datetime')

print("All checks passed")
```
